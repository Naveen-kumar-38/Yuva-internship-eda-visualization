# Week 2: Exploratory Data Analysis and Visualization

## Objective

The objective of this task is to perform Exploratory Data Analysis (EDA) on the cleaned automobile dataset obtained during Week 1. The analysis focuses on understanding the structure, distribution, relationships, trends, patterns, and possible anomalies present in the dataset.

Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn are used to perform statistical analysis and create meaningful visualizations.

## 1. Import Required Libraries

The required Python libraries are imported for data manipulation, statistical analysis, and visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from IPython.display import display

sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

## 2. Define File Paths

The dataset, output, and screenshot directories are defined so that the analysis results and visualizations can be stored systematically.

```python
DATA_PATH = r"D:\Projects\Yuva_Intern\Task_2_EDA\data\cleaned_automobile.csv"
OUTPUT_DIR = r"D:\Projects\Yuva_Intern\Task_2_EDA\output"
SCREENSHOT_DIR = r"D:\Projects\Yuva_Intern\Task_2_EDA\screenshots"

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(SCREENSHOT_DIR, exist_ok=True)

print("Paths configured successfully.")
```

## 3. Load the Cleaned Dataset

The cleaned automobile dataset produced during Week 1 is loaded using Pandas. The first few records are displayed to verify that the dataset has been loaded correctly.

```python
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Dataset Shape:", df.shape)

display(df.head())
```

## 4. Basic Dataset Information

The dimensions, column names, data types, and missing-value status of the dataset are examined before performing detailed exploratory analysis.

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

## 5. Statistical Summary

Descriptive statistics are calculated for the numerical variables. Measures such as mean, standard deviation, minimum, maximum, and quartiles help in understanding the distribution and variation of automobile characteristics.

```python
print("Statistical Summary:")
display(df.describe().T)
```

## 6. Numerical and Categorical Variables

The dataset contains both numerical and categorical variables. Numerical variables are useful for statistical and correlation analysis, while categorical variables are examined using frequency-based analysis and visualizations.

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(include="object").columns.tolist()

print("Numerical Columns:")
print(numeric_cols)

print("\nCategorical Columns:")
print(categorical_cols)

print("\nNumber of Numerical Columns:", len(numeric_cols))
print("Number of Categorical Columns:", len(categorical_cols))
```

## 7. Price Distribution

The distribution of automobile prices is analyzed using a histogram with a density curve. This helps identify the concentration of prices and whether the distribution is symmetric or skewed.

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["price"], kde=True, bins=30)

plt.title("Distribution of Automobile Prices")
plt.xlabel("Price")
plt.ylabel("Number of Cars")

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "02_price_distribution.png"),
    dpi=300
)

plt.show()
```

### Observation

The price distribution shows that most automobiles are concentrated within the lower-to-middle price range, while a smaller number of vehicles have considerably higher prices.

### Interpretation

The distribution suggests that automobile prices are not evenly distributed. The presence of higher-priced vehicles creates a longer upper range, indicating variation in the market segments represented in the dataset.

## 8. Distribution of Automobile Makes

The number of automobiles belonging to each manufacturer is analyzed to understand how the dataset is distributed across different automobile makes.

```python
plt.figure(figsize=(12, 6))

make_counts = df["make"].value_counts()

sns.barplot(
    x=make_counts.index,
    y=make_counts.values
)

plt.title("Number of Automobiles by Make")
plt.xlabel("Automobile Make")
plt.ylabel("Number of Cars")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "03_automobile_makes.png"),
    dpi=300
)

plt.show()
```

### Observation

The dataset contains automobiles from several manufacturers, but the number of records differs between manufacturers.

### Interpretation

This indicates that the dataset is not evenly distributed across all automobile makes. Some manufacturers have a larger representation than others.

## 9. Fuel Type Distribution

The frequency of different fuel types is analyzed to determine which fuel type is most commonly represented in the dataset.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="fuel_type")

plt.title("Distribution of Fuel Types")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "04_fuel_type.png"),
    dpi=300
)

plt.show()
```

### Observation

Gasoline-powered automobiles form the majority of the dataset, while diesel automobiles represent a smaller proportion.

### Interpretation

The distribution indicates that gasoline vehicles are more strongly represented in this particular dataset.

## 10. Automobile Body Style Distribution

The distribution of different automobile body styles is analyzed to identify the most frequently represented vehicle categories.

```python
plt.figure(figsize=(9, 5))

body_counts = df["body_style"].value_counts()

sns.barplot(
    x=body_counts.index,
    y=body_counts.values
)

plt.title("Distribution of Automobile Body Styles")
plt.xlabel("Body Style")
plt.ylabel("Number of Cars")

plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "05_body_style.png"),
    dpi=300
)

plt.show()
```

### Observation

The dataset contains multiple automobile body styles, with some categories appearing more frequently than others.

### Interpretation

The unequal distribution shows that certain vehicle body styles have greater representation in the dataset.

## 11. Horsepower Distribution

The distribution of horsepower values is analyzed to understand the typical performance range of the automobiles in the dataset.

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["horsepower"], kde=True, bins=25)

plt.title("Distribution of Automobile Horsepower")
plt.xlabel("Horsepower")
plt.ylabel("Number of Cars")

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "06_horsepower_distribution.png"),
    dpi=300
)

plt.show()
```

### Observation

Horsepower values are concentrated around the lower and middle ranges, while fewer automobiles have very high horsepower.

### Interpretation

The distribution indicates that high-performance vehicles form a smaller portion of the dataset compared with vehicles having moderate horsepower.

## 12. Engine Size and Automobile Price

A scatter plot is used to investigate the relationship between engine size and automobile price.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="engine_size",
    y="price"
)

plt.title("Relationship Between Engine Size and Price")
plt.xlabel("Engine Size")
plt.ylabel("Price")

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "07_engine_size_vs_price.png"),
    dpi=300
)

plt.show()
```

### Observation

The scatter plot shows a generally positive relationship between engine size and automobile price.

### Interpretation

Automobiles with larger engines tend to appear at higher price levels. However, engine size alone does not determine price because other factors such as horsepower, make, body style, and equipment can also influence automobile prices.

## 13. Horsepower and Automobile Price

The relationship between horsepower and automobile price is examined using a scatter plot.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="horsepower",
    y="price"
)

plt.title("Relationship Between Horsepower and Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "08_horsepower_vs_price.png"),
    dpi=300
)

plt.show()
```

### Observation

Higher-horsepower automobiles generally tend to appear at higher price levels.

### Interpretation

The positive relationship suggests that performance-related characteristics can be associated with automobile price. However, the relationship is not perfectly linear because multiple factors influence vehicle pricing.

## 14. City Mileage and Automobile Price

The relationship between city fuel efficiency and automobile price is examined using a scatter plot.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="city_mpg",
    y="price"
)

plt.title("Relationship Between City MPG and Price")
plt.xlabel("City MPG")
plt.ylabel("Price")

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "09_city_mpg_vs_price.png"),
    dpi=300
)

plt.show()
```

## 15. Automobile Price by Body Style

A boxplot is used to compare automobile price distributions across different body styles. The visualization helps identify differences in median prices, variation, and potential extreme observations.

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="body_style",
    y="price"
)

plt.title("Automobile Price by Body Style")
plt.xlabel("Body Style")
plt.ylabel("Price")

plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "10_price_by_body_style.png"),
    dpi=300
)

plt.show()
```

### Observation

The price ranges vary across different body styles. The boxplot also shows differences in median prices and the spread of prices within each category.

### Interpretation

Body style can be associated with differences in automobile pricing, although it should not be considered independently because other vehicle characteristics also affect price.

## 16. Correlation Analysis

Correlation analysis is performed on the numerical variables to identify relationships between automobile characteristics. A heatmap is used to make the relationships easier to interpret visually.

```python
correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")

plt.tight_layout()

plt.savefig(
    os.path.join(SCREENSHOT_DIR, "11_correlation_heatmap.png"),
    dpi=300
)

plt.show()
```

## 17. Correlation of Numerical Variables with Price

The correlation of each numerical variable with automobile price is calculated to identify which numerical characteristics have stronger relationships with price.

```python
corr_price = correlation_matrix["price"].sort_values(ascending=False)

print("Correlation of Numerical Features with Price:")
display(corr_price.to_frame(name="Correlation with Price"))
```

## 18. Strongest Numerical Relationships

The strongest positive and negative correlations between numerical variables are identified. This helps highlight variables that show similar or opposite patterns in the dataset.

```python
corr_pairs = correlation_matrix.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) != corr_pairs.index.get_level_values(1)
]

corr_pairs = corr_pairs.sort_values(ascending=False)

print("Strongest Positive Relationships:")
display(corr_pairs.head(10))

print("\nStrongest Negative Relationships:")
display(corr_pairs.tail(10))
```

## 19. Overall EDA Summary

The major statistical characteristics and categorical patterns identified during the exploratory analysis are summarized below.

```python
print("========== EDA SUMMARY ==========\n")

print("Dataset Size:", df.shape)

print("\nPrice Statistics:")
print("Mean Price:", round(df["price"].mean(), 2))
print("Median Price:", round(df["price"].median(), 2))
print("Minimum Price:", df["price"].min())
print("Maximum Price:", df["price"].max())

print("\nMost Common Automobile Make:")
print(df["make"].value_counts().idxmax())

print("\nMost Common Body Style:")
print(df["body_style"].value_counts().idxmax())

print("\nMost Common Fuel Type:")
print(df["fuel_type"].value_counts().idxmax())

print("\nAverage Horsepower:")
print(round(df["horsepower"].mean(), 2))

print("\nAverage City MPG:")
print(round(df["city_mpg"].mean(), 2))

print("\nAverage Highway MPG:")
print(round(df["highway_mpg"].mean(), 2))
```

## 20. Save EDA Summary

The important numerical and categorical findings are stored in a CSV file for documentation and future reference.

```python
summary = pd.DataFrame({
    "Metric": [
        "Number of Rows",
        "Number of Columns",
        "Mean Price",
        "Median Price",
        "Minimum Price",
        "Maximum Price",
        "Most Common Make",
        "Most Common Body Style",
        "Most Common Fuel Type",
        "Average Horsepower",
        "Average City MPG",
        "Average Highway MPG"
    ],
    "Value": [
        df.shape[0],
        df.shape[1],
        round(df["price"].mean(), 2),
        round(df["price"].median(), 2),
        df["price"].min(),
        df["price"].max(),
        df["make"].value_counts().idxmax(),
        df["body_style"].value_counts().idxmax(),
        df["fuel_type"].value_counts().idxmax(),
        round(df["horsepower"].mean(), 2),
        round(df["city_mpg"].mean(), 2),
        round(df["highway_mpg"].mean(), 2)
    ]
})

summary.to_csv(
    os.path.join(OUTPUT_DIR, "eda_summary.csv"),
    index=False
)

display(summary)

print("\nEDA summary saved successfully.")
```

The final dataset is checked to confirm its dimensions, missing values, and duplicate records after completing the exploratory analysis.

```python
print("========== FINAL VALIDATION ==========")

print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

print("\nNumerical Columns:", len(numeric_cols))
print("Categorical Columns:", len(categorical_cols))

print("\nEDA analysis completed successfully.")
```
